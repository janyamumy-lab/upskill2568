P**andas (แพนดาส)** เป็นไลบรารี (Library) ของภาษา Python ที่ได้รับความนิยมสูงสุดสำหรับการวิเคราะห์ จัดการ และเตรียมข้อมูล (Data Manipulation & Analysis) โดยเฉพาะข้อมูลรูปแบบตาราง (Tabular Data) เช่น CSV หรือ SQL โดยมีโครงสร้างหลักคือ DataFrame ที่ช่วยให้ทำความสะอาด แปลงข้อมูล และจัดทำรายงานได้รวดเร็วเพียงไม่กี่บรรทัด

**ความสามารถหลักของ Pandas:**
*   จัดการข้อมูล (Data Manipulation): อ่าน, เขียน, จัดการข้อมูลในรูปแบบตาราง (DataFrame) และแถวลำดับ (Series) ได้อย่างง่ายดาย
*   ทำความสะอาดข้อมูล (Data Cleaning): จัดการข้อมูลสูญหาย (Missing Data), เปลี่ยนประเภทข้อมูล, ลบข้อมูลซ้ำ หรือกรองข้อมูลที่ไม่ต้องการ
*   การวิเคราะห์ขั้นสูง (Data Analysis): รองรับการแบ่งกลุ่ม (GroupBy), การเชื่อมตาราง (Merge/Join), การเรียงลำดับ และการทำ Pivoting
*   ทำงานร่วมกับ Library อื่น: ใช้ร่วมกับ NumPy (คำนวณเชิงตัวเลข) และ Matplotlib/Seaborn (วาดกราฟ) ได้อย่างลงตัว

**จุดเด่นและหน้าที่หลักของ Matplotlib:**

*   ปรับแต่งได้ละเอียด: สามารถปรับแต่งสี, ขนาด, เส้น, ฟอนต์ และองค์ประกอบต่างๆ ของกราฟได้ตามต้องการ
*   ใช้งานร่วมกับ Python Ecosystem: ทำงานได้ดีร่วมกับ Pandas และ NumPy
*   เป็นโอเพนซอร์ส (Open Source): ใช้งานได้ฟรีและได้รับความนิยมสูง

Matplotlib ช่วยเปลี่ยนข้อมูลเชิงตัวเลขให้เป็นภาพที่เข้าใจง่ายและสวยงาม เหมาะสำหรับการสรุปข้อมูลและการนำเสนอผลการวิเคราะห์ข้อมูล


```python
import pandas as pd
import matplotlib.pyplot as plt

#โหลด dataset
df = pd.read_csv('https://raw.githubusercontent.com/janyamumy-lab/upskill2568/refs/heads/main/retail_sales_dataset.csv')

#ขั้นที่ 1: ดูโครงสร้างข้อมูล
print("Shape: ")
print(df.shape)   #จำนวนแถวและคอลัมน์
print("\nFirst 5 rows :")
print(df.head())         #แสดงตัวอย่างข้อมูล 5 แถวแรก
print("\nData Info:")
print(df.info())         #แสดงรายละเอียดคอลัมน์ ชนิดข้อมูล missing

```

```python
#ขั้นที่ 2: สถิติเบื้องต้น
print("\nData Description:")
print(round(df.describe(),2))     #Mean, Max, Min, Std, Percentile...
print("\nChecking missing value:")
print(df.isnull().sum()) #จำนวน missing value
```

```python
#สำรวจ Category
df['Category'].value_counts() #ค่าของข้อมูล
```

**Build Visualization**

Data Visualization คือ กระบวนการแปลงข้อมูล (Data) ที่ซับซ้อน ตัวเลขจำนวนมาก หรือข้อความ ให้กลายเป็นภาพที่มองเห็นได้ (Visuals) เช่น กราฟ แผนภูมิ แผนที่ อินโฟกราฟิก หรือแดชบอร์ด เพื่อเปลี่ยนข้อมูลที่เข้าใจยากให้เป็นเรื่องที่เข้าใจง่าย มองเห็นแนวโน้ม (Trends) รูปแบบ (Patterns) หรือความสัมพันธ์ของข้อมูลได้อย่างรวดเร็วและชัดเจน

**สาระสำคัญของ Visualization**
*   แปลงข้อมูลเป็นภาพ: ปลี่ยนข้อมูลเชิงปริมาณหรือข้อความให้เป็นแผนภูมิรูปภาพ
*   เข้าใจง่ายและรวดเร็ว: ช่วยให้ผู้ใช้งานประหยัดเวลาในการวิเคราะห์ข้อมูลและมองเห็นภาพรวมได้อย่างชัดเจน
*   ช่วยตัดสินใจ: ช่วยให้ผู้บริหารหรือผู้ใช้งานตัดสินใจได้อย่างแม่นยำและชาญฉลาดขึ้น
*   ประโยชน์หลัก: ทำให้ข้อมูลน่าสนใจ จดจำได้ง่ายกว่าการดูตารางตัวเลข

**ตัวอย่างการทำ Visualization (ประเภทของกราฟ)**
*   กราฟแท่ง (Bar Chart): เปรียบเทียบข้อมูลแต่ละประเภท
*   กราฟเส้น (Line Chart): แสดงแนวโน้มของข้อมูลตามช่วงเวลา
*   แผนภูมิวงกลม (Pie Chart): แสดงส่วนแบ่งข้อมูลเป็นเปอร์เซ็นต์
*   แผนที่ (Map): แสดงข้อมูลเชิงภูมิศาสตร์
*   แดชบอร์ด (Dashboard): รวบรวมกราฟหลายๆ ประเภทมาไว้ในที่เดียวเพื่อสรุปผลการดำเนินงาน

**เครื่องมือที่นิยมใช้ (Visualization Tools)**

โดยสรุปแล้ว การทำ Data Visualization เปรียบเสมือนการนำข้อมูลมาเล่าเรื่องผ่านภาพ เพื่อให้เห็นสิ่งที่ซ่อนอยู่และนำไปใช้งานต่อได้จริง


```python
#แสดงยอดรวมของ Total_Amount
df_sales = df['Total_Amount'].sum()
print("Total Amount : ",df_sales)
#แสดงยอดขายเฉลี่ย
df_avg = df['Total_Amount'].mean()
print("\nAverage Amount : ",df_avg)
#แสดง Category ที่มียอดขายสูงสูด
df_max = df.loc[df['Total_Amount'].idxmax()]
print("\nMax Amount : ",df_max)
```

```python
#Bar Chart : ยอดขายแยกตาม Category
df_cat = df.groupby('Category')['Total_Amount'].sum()
df_cat.plot(kind='bar', color=['lightblue', 'lightgreen', 'orange'])
#แสดง value ของแต่ละ category
for i in range(len(df_cat)):
  plt.text(i, df_cat[i], df_cat[i], ha='center', va='bottom')

plt.title('Total Amount by Category')
plt.xlabel('Category')
plt.ylabel('Total Amount')
```

```python
#Bar Chart : จำนวน Transaction แยกตาม Category
df_trans = df.groupby('Category')['Transaction ID'].count()
df_trans.plot(kind='bar', color=['lightblue', 'lightgreen', 'orange'])
#แสดง value ของแต่ละ category
for i in range(len(df_trans)):
  plt.text(i, df_trans[i], df_trans[i], ha='center', va='bottom')

plt.title('Freequency by Category')
plt.xlabel('Category')
plt.ylabel('Frequency')
```

```python
#Pie chart: สัดส่วนจำนวนสินค้าที่ขายได้แยกตาม Category
df_qty = df.groupby('Category')['Quantity'].sum()
df_qty.plot(kind='pie', autopct='%1.1f%%',colors=['lightblue', 'lightgreen', 'orange'])
plt.title('Quantity by Category')
plt.ylabel('')
plt.show()

#Bar chart
df_qty.plot(kind='bar', color=['lightblue', 'lightgreen', 'orange'])
#แสดง value ของแต่ละ category
for i in range(len(df_qty)):
  plt.text(i, df_qty[i], df_qty[i], ha='center', va='bottom')

plt.xlabel('Category')
plt.ylabel('Sum of Quantity')
```

```python
#Line Chart: แนวโน้มรายเดือน
df['Month'] = pd.to_datetime(df['Date']).dt.month
monthly = df.groupby('Month')['Total_Amount'].sum()
monthly.plot(kind='line', marker='o')
#แสดงชื่อเดือนแกน x เป็นตัวอักษรอย่างย่อ
monthly_text = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
plt.xticks(range(1, 13), monthly_text)
plt.ylabel('Total Amount')
plt.title('Monthly Total Amount')
plt.show()
```

```python
from numpy._core.defchararray import title
#Scatter: ราคา vs จำนวนที่ขาย
import seaborn as sns
sns.scatterplot(x='UnitPrice', y='Quantity', data=df, hue='Category')
plt.xlabel('Unit Price')
plt.ylabel('Quantity')
plt.title('Price vs Quantity')
```
